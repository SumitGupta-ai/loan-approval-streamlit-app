## Feature  Engineering and Selection 

```python
x = loan.drop('loan_status', axis=1)
y = loan['loan_status']
```

## Create column-Transformer and Pipeline

```python
# Only Numeric Feature
num_features = ['person_age', 'person_income', 'loan_amnt', 'loan_int_rate', 
                'loan_percent_income','credit_score']
# Binary Feature
bin_features = ['person_gender', 'previous_loan_defaults_on_file']
# Categorical Feature
cat_feature = ['loan_intent']
```

```python
# Column Transformer 
preprocess = ColumnTransformer(
transformers=[('num', StandardScaler(), num_features),
              ('bin', OrdinalEncoder(), bin_features),
                ('cat', OneHotEncoder(handle_unknown='ignore'), cat_feature)]
)
```

### Train Test Split

```python
# Train Data 80% And Test Data 20%
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
x_train.head(1)
```

```python
print(x_train.columns.tolist())
```

```python
# Create a multiple Models
models = {
    "Logistic Regression": LogisticRegression(),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(random_state=42),
    'SVM' : SVC()
}
```

## Evaluates models

```python
# Create Pipeline and train dataset and predict test data and check accuracy and confusiton matrix and classification report
results = []

for name, model in models.items():
    pipe = Pipeline([
        ('preprocess', preprocess),
        ('model', model)
    ])
    pipe.fit(x_train, y_train)
    preds = pipe.predict(x_test)
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, preds) })
    # print(name, 'accuracy', accuracy_score(y_test, preds))
    print(name, 'confusion matrix\n', confusion_matrix(y_test, preds))
    print(name, 'Classification report\n', classification_report(y_test, preds))
```

```python
df_results = pd.DataFrame(results)
print(df_results)
```

```python
plt.figure(figsize=(6, 6))
sns.barplot(data=df_results, x='Model', y='Accuracy')

plt.title('Accuracy Comparison of Models')
plt.ylim(0, 1)
plt.xticks(rotation=45)
plt.show()
```

```python
# Select the final model RandomForestClassifier
Final_Select_model = Pipeline([('preprcessing', preprocess),
                ('model', RandomForestClassifier())])                         
```

```python
# Train model
Final_Select_model.fit(x_train, y_train)
```

```python
y_proba = Final_Select_model.predict_proba(x_test)[:,1]
```

```python
from sklearn.metrics import roc_auc_score
```

```python
roc_auc = roc_auc_score(y_test, y_proba)
print('ROC-AUC Score:', roc_auc)
```

```python
from sklearn.metrics import roc_curve
fpr, tpr, thresholds = roc_curve(y_test, y_proba)

plt.figure()
plt.plot(fpr, tpr)
plt.plot([0, 1], [0, 1], linestyle='--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.show()
```
